## Cleaning
Run `src/data_cleaning.py` from the project root to create the cleaned dataset.

```python
clean = pd.read_csv('../output/cleaned_students.csv')
clean.head()
```

```python
subjects = ['python','dsa','dbms','maths','os']
print('Average CGPA:', clean['cgpa'].mean())
print('Placement rate:', clean['placed'].eq('yes').mean()*100)
print(clean[subjects].mean())
```

```python
clean.groupby('placed')[['cgpa','attendance','projects','internships','backlogs']].mean().round(2)
```

```python
plt.figure(figsize=(7,4))
plt.scatter(clean['attendance'], clean['cgpa'], alpha=0.5)
plt.xlabel('Attendance (%)')
plt.ylabel('CGPA')
plt.title('Attendance vs CGPA')
plt.show()
```
